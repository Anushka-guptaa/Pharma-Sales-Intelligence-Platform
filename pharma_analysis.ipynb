# Visual Exploratory Data Analysis (EDA)

```python
"""Now we'll visualize the data.

Why?

Because our brain understands graphs much faster than tables."""
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
country_sales = df.groupby("Country")["Sales"].sum()

country_sales.plot(kind="bar", figsize=(6,4))
plt.title("Total Sales by Country")
plt.xlabel("Country")
plt.ylabel("Sales")
plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Products by Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
""" 10 inches wide and 5 inches tall (10,5)
We know the products are on the horizontal bottom line (the X-axis) 
because of a specific rule built into Python's pandas: the names you group your data by 
automatically become the labels on your chart's baseline.
"""
```

```python
channel_sales = df.groupby("Channel")["Sales"].sum()

channel_sales.plot(kind="bar")

plt.title("Sales by Channel")
plt.ylabel("Sales")

plt.show()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")

plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.hist(df["Sales"], bins=30)

plt.title("Distribution of Sales")

plt.xlabel("Sales")

plt.ylabel("Frequency")

plt.show()
```

```python
""" A histogram groups numerical values into ranges called **bins** and shows how many values fall within each range. 
For example, if sales values are **100, 200, 300, 500, 700, and 50,00,000**, using **10 bins** creates large ranges,
so all the smaller sales (100–700) may fall into the same bin, making them appear as one large bar and hiding the actual distribution. 
Using **30 bins** creates smaller ranges, allowing the smaller sales to spread across multiple bins and making the pattern easier to understand.
However, using **1000 bins** creates extremely small ranges, so many bins contain no values and the bars become too thin to interpret. 
Therefore, the number of bins should be chosen to provide enough detail while keeping the histogram clear and readable."""
```
